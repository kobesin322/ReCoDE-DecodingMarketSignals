# QQQ 200 SMA Pyramid Accumulation vs Monthly DCA

**Strategy Overview:**
- Uses selective 200 SMA breakout as buy signals (full close above + next-day open confirmation)
- Each signal adds $10,000 position (max 12 per year / total $120k)
- No stop loss, no exits — hold all positions indefinitely (pyramid style)
- Compare final value vs DCA ($10k monthly, same total capital)

**Ticker:** QQQ (Invesco QQQ Trust ETF)


```python
# Imports
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

```python
# Data Download
ticker = 'QQQ'

print(f"Downloading {ticker} data...")

# Download QQQ data from inception (~1999)
data = yf.download(ticker, start='1999-01-01', end=datetime.now().strftime('%Y-%m-%d'), progress=False, auto_adjust=True)

# Handle MultiIndex columns (yfinance returns MultiIndex for single ticker)
if isinstance(data.columns, pd.MultiIndex):
    # Flatten MultiIndex columns
    data.columns = data.columns.get_level_values(0)

# Select required columns
required_cols = ['Open', 'High', 'Low', 'Close', 'Volume']
if 'Adj Close' in data.columns:
    required_cols.append('Adj Close')

data = data[required_cols].copy()

# If Adj Close doesn't exist, create it from Close
if 'Adj Close' not in data.columns:
    data['Adj Close'] = data['Close']

print(f"\nDownloaded {len(data)} records")
print(f"Date range: {data.index[0]} to {data.index[-1]}")
print(f"Columns: {list(data.columns)}")
print("\nLast few rows:")
print(data.tail())
```

```python
# Calculate Indicators & Entry Signals

# Calculate 200-day Simple Moving Average
data['SMA200'] = data['Close'].rolling(window=200, min_periods=200).mean()

# Breakout signal: Close > SMA200 AND previous Close <= SMA200 (breakout detected)
data['Breakout'] = (data['Close'] > data['SMA200']) & (data['Close'].shift(1) <= data['SMA200'].shift(1))

# Next day entry signal: If breakout detected yesterday, check if today's Open > SMA200
data['NextDay_Entry'] = False
data.loc[data['Breakout'].shift(1) == True, 'NextDay_Entry'] = (data['Open'] > data['SMA200'])

# Drop rows where SMA200 is NaN (first 200 days)
data = data.dropna(subset=['SMA200'])

# Add year column for entry limiting
data['Year'] = data.index.year

# Limit to max 12 entries per year
max_entries_per_year = 12
entry_counts = data[data['NextDay_Entry']].groupby('Year').size()

# Track entries per year to enforce limit
data['Entry_Limit_OK'] = True
yearly_entry_count = {}
for year in data['Year'].unique():
    yearly_entry_count[year] = 0

# Mark entries that exceed the limit
for date, row in data.iterrows():
    if row['NextDay_Entry']:
        year = row['Year']
        if yearly_entry_count[year] >= max_entries_per_year:
            data.loc[date, 'NextDay_Entry'] = False
        else:
            yearly_entry_count[year] += 1

# Print statistics
print(f"Total potential entries (before limit): {(data['Breakout'].shift(1) == True).sum()}")
print(f"Total entries after next-day filter: {data['NextDay_Entry'].sum()}")
print(f"\nEntries per year (after limit):")
final_entry_counts = data[data['NextDay_Entry']].groupby('Year').size()
for year, count in final_entry_counts.items():
    if count > 0:
        print(f"  {year}: {count} entries")
    if count > max_entries_per_year:
        print(f"    Warning: {year} exceeds {max_entries_per_year} limit!")

print(f"\nData ready for simulation: {len(data)} records")
print(f"Date range: {data.index[0]} to {data.index[-1]}")
```

```python
# Simulate Pyramid Strategy

# Strategy parameters (tunable)
invest_amount_per_signal = 10000  # $10,000 per entry signal
max_total_invested = 120000  # Maximum total investment ($120k)

# Initialize tracking variables
total_invested = 0
positions = []  # list of (entry_date, entry_price, shares)
portfolio_value = []
invested_series = []
entry_dates_list = []

print("Simulating Pyramid Strategy...")
print("="*60)

# Loop through each trading day
for date, row in data.iterrows():
    # Calculate current portfolio value (sum of all positions at current Adj Close)
    current_value = sum(shares * row['Adj Close'] for _, _, shares in positions)
    portfolio_value.append(current_value)
    invested_series.append(total_invested)
    
    # Check for entry signal
    if row['NextDay_Entry'] and total_invested < max_total_invested:
        # Calculate investment amount (may be less if near limit)
        invest_amount = min(invest_amount_per_signal, max_total_invested - total_invested)
        
        if invest_amount > 0:
            # Buy at opening price
            entry_price = row['Open']
            shares = invest_amount / entry_price
            
            # Add position
            positions.append((date, entry_price, shares))
            total_invested += invest_amount
            entry_dates_list.append(date)
            
            print(f"Entry {len(positions)}: {date.strftime('%Y-%m-%d')} @ ${entry_price:.2f}, Shares: {shares:.2f}, Invested: ${invest_amount:,.2f}")

# Create series for plotting
strategy_value = pd.Series(portfolio_value, index=data.index[:len(portfolio_value)])
invested_series = pd.Series(invested_series, index=data.index[:len(invested_series)])

# Calculate final metrics
final_strategy_value = strategy_value.iloc[-1]
strategy_return_ratio = final_strategy_value / total_invested if total_invested > 0 else 0

print(f"\n{'='*60}")
print("PYRAMID STRATEGY RESULTS")
print(f"{'='*60}")
print(f"Total Entries: {len(positions)}")
print(f"Total Invested: ${total_invested:,.2f}")
print(f"Final Portfolio Value: ${final_strategy_value:,.2f}")
print(f"Return Multiple: {strategy_return_ratio:.2f}x")
print(f"Total Return: {(strategy_return_ratio - 1) * 100:.2f}%")
print(f"{'='*60}")
```

```python
# Simulate DCA Benchmark

# DCA parameters
dca_invest_per_month = 10000  # $10,000 per month
max_total_dca_invested = 120000  # Maximum total investment ($120k)

# Initialize tracking variables
total_dca_invested = 0
dca_shares = 0
dca_values = []
dca_dates = []
dca_investment_count = 0

print("Simulating Monthly DCA Benchmark...")
print("="*60)

# Resample to monthly data (end of month)
monthly_data = data.resample('M').last()

# Loop through each month
for date, row in monthly_data.iterrows():
    # Invest first, then calculate value
    if total_dca_invested < max_total_dca_invested:
        # Calculate investment amount (may be less if near limit)
        invest = min(dca_invest_per_month, max_total_dca_invested - total_dca_invested)
        
        if invest > 0:
            # Buy at month-end Adj Close price
            shares = invest / row['Adj Close']
            dca_shares += shares
            total_dca_invested += invest
            dca_investment_count += 1
            
            print(f"DCA Entry {dca_investment_count}: {date.strftime('%Y-%m-%d')} @ ${row['Adj Close']:.2f}, Shares: {shares:.2f}, Invested: ${invest:,.2f}")
    
    # Calculate current DCA portfolio value (after potential investment)
    current_dca_value = dca_shares * row['Adj Close']
    dca_values.append(current_dca_value)
    dca_dates.append(date)

# Create series for plotting
dca_series = pd.Series(dca_values, index=dca_dates[:len(dca_values)])

# Calculate final metrics
final_dca_value = dca_series.iloc[-1]
dca_return_ratio = final_dca_value / total_dca_invested if total_dca_invested > 0 else 0

print(f"\n{'='*60}")
print("DCA BENCHMARK RESULTS")
print(f"{'='*60}")
print(f"Total Monthly Investments: {dca_investment_count}")
print(f"Total Invested: ${total_dca_invested:,.2f}")
print(f"Final Portfolio Value: ${final_dca_value:,.2f}")
print(f"Return Multiple: {dca_return_ratio:.2f}x")
print(f"Total Return: {(dca_return_ratio - 1) * 100:.2f}%")
print(f"{'='*60}")
```

```python
# Comparison Plot & Table

# Create comparison plot
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10))

# Plot 1: Portfolio Value Over Time
strategy_value.plot(ax=ax1, label='Pyramid Strategy (on signals)', linewidth=2, color='#16A085', alpha=0.8)
# Reindex DCA series to match strategy index for comparison
dca_aligned = dca_series.reindex(strategy_value.index, method='ffill')
dca_aligned.plot(ax=ax1, label='Monthly DCA', linewidth=2, color='#F77F00', alpha=0.8)

# Mark entry points for pyramid strategy
if len(entry_dates_list) > 0:
    entry_values = [strategy_value.loc[date] if date in strategy_value.index else 0 for date in entry_dates_list]
    ax1.scatter(entry_dates_list, entry_values, color='green', marker='^', s=100, alpha=0.6, label='Pyramid Entries', zorder=5)

ax1.set_title(f'{ticker} - Pyramid vs DCA Comparison\nFinal Multiple: Strategy {strategy_return_ratio:.2f}x vs DCA {dca_return_ratio:.2f}x', 
              fontsize=14, fontweight='bold')
ax1.set_ylabel('Portfolio Value ($)', fontsize=12)
ax1.legend(loc='best', fontsize=10)
ax1.grid(True, alpha=0.3)

# Plot 2: Cumulative Investment Over Time
invested_series.plot(ax=ax2, label='Pyramid Strategy Invested', linewidth=2, color='#16A085', alpha=0.8, linestyle='--')

# Calculate DCA cumulative investment series
dca_invested_values = []
dca_cumulative = 0
for date, row in monthly_data.iterrows():
    if dca_cumulative < max_total_dca_invested:
        invest = min(dca_invest_per_month, max_total_dca_invested - dca_cumulative)
        dca_cumulative += invest
    dca_invested_values.append(dca_cumulative)
dca_invested_series = pd.Series(dca_invested_values, index=monthly_data.index[:len(dca_invested_values)])
dca_invested_series.plot(ax=ax2, label='DCA Invested', linewidth=2, color='#F77F00', alpha=0.8, linestyle='--')
ax2.set_title('Cumulative Investment Over Time', fontsize=12, fontweight='bold')
ax2.set_xlabel('Date', fontsize=12)
ax2.set_ylabel('Cumulative Investment ($)', fontsize=12)
ax2.legend(loc='best', fontsize=10)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Create comparison table
comparison = pd.DataFrame({
    'Metric': [
        'Final Portfolio Value',
        'Total Invested',
        'Return Multiple',
        'Total Return (%)',
        'Number of Entries'
    ],
    'Pyramid Strategy': [
        f"${final_strategy_value:,.0f}",
        f"${total_invested:,.0f}",
        f"{strategy_return_ratio:.2f}x",
        f"{(strategy_return_ratio - 1) * 100:.2f}%",
        f"{len(positions)}"
    ],
    'Monthly DCA': [
        f"${final_dca_value:,.0f}",
        f"${total_dca_invested:,.0f}",
        f"{dca_return_ratio:.2f}x",
        f"{(dca_return_ratio - 1) * 100:.2f}%",
        f"{dca_investment_count}"
    ]
})

print(f"\n{'='*60}")
print("STRATEGY COMPARISON")
print(f"{'='*60}")
print(comparison.to_string(index=False))
print(f"{'='*60}")

# Calculate difference
value_difference = final_strategy_value - final_dca_value
ratio_difference = strategy_return_ratio - dca_return_ratio

print(f"\nDifference:")
print(f"  Value Difference: ${value_difference:,.2f} ({'Pyramid' if value_difference > 0 else 'DCA'} is higher)")
print(f"  Return Multiple Difference: {ratio_difference:.2f}x ({'Pyramid' if ratio_difference > 0 else 'DCA'} is better)")
print(f"{'='*60}")
```

## Conclusion

**Pyramid Strategy:**
- Invests only on 200 SMA breakout signals (max $120k total)
- Selective entry: Only when price fully closes above 200 SMA, then confirms on next day's open
- Maximum 12 entries per calendar year
- No stop loss, no exits — all positions held indefinitely (pyramid accumulation)
- Final value calculated using Adj Close to account for dividends/splits

**Monthly DCA Benchmark:**
- Invests $10,000 every month regardless of price
- Same total capital ($120,000) over the same period
- Final value calculated using Adj Close

**Comparison Metric:**
- **Return Multiple** = Final Portfolio Value / Total Invested
- Higher multiple indicates better capital efficiency
- Compare which approach gives higher final amount per dollar invested

**Tunable Parameters:**
- `invest_amount_per_signal`: Investment per entry signal (default: $10,000)
- `max_total_invested`: Maximum total investment (default: $120,000)
- `max_entries_per_year`: Maximum entries per calendar year (default: 12)
- `dca_invest_per_month`: Monthly DCA investment (default: $10,000)
